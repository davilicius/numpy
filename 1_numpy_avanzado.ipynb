```python
import random
```

```python
def generar_matriz(N):
    matriz= []
    for i in range (N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1, 9))
        matriz.append(fila)
    return matriz

N = 5

matriz_aleatoria = generar_matriz(N)
for fila in matriz_aleatoria:
    print(fila)

def imprimir_matriz(matriz):
    for fila in matriz:
        fila_formateada = ' '.join(f'{elemento:2}' for elemento in fila)
        print(fila_formateada)
```

```python
print("Matriz de 5x5")
m_5 = generar_matriz(5)
imprimir_matriz(m_5)
print("============")
print("Matriz de 10x10")
m_10 = generar_matriz(10)
imprimir_matriz(m_10)
print("============")
print("Matriz de 20x20")
m_20 = generar_matriz(20)
imprimir_matriz(m_20)
```

```python
assert array_1d.shape == (3, 4)
assert array_1d.ndim == 2 and array_1d.size == 12
for n, m in [(5, m_5), (10, m_10), (20, m_20)]:
    assert np.asarray(m).shape == (n, n)
    assert ((np.asarray(m) >= 1) & (np.asarray(m) <= 9)).all()
print('Verificado: arreglo 3x4 con 12 elementos y matrices de 5x5, 10x10 y 20x20.')
```
